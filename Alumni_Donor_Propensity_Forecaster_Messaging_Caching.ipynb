```python
# COMMAND: Create RabbitMQ connection

def create_rabbitmq_connection():

    credentials = pika.PlainCredentials(
        RMQ_USERNAME,
        RMQ_PASSWORD
    )

    parameters = pika.ConnectionParameters(
        host=RMQ_HOST,
        port=RMQ_PORT,
        virtual_host=RMQ_VHOST,
        credentials=credentials
    )

    return pika.BlockingConnection(parameters)

print("✅ RabbitMQ connection function created!")
```

```python
# COMMAND: Create RabbitMQ producer

def publish_prediction(result):

    connection = create_rabbitmq_connection()

    channel = connection.channel()

    channel.queue_declare(
        queue=QUEUE_NAME,
        durable=True
    )

    event = {
        "event": "donor_prediction_created",
        "data": result
    }

    channel.basic_publish(
        exchange="",
        routing_key=QUEUE_NAME,
        body=json.dumps(event),
        properties=pika.BasicProperties(
            delivery_mode=2
        )
    )

    print("📤 Prediction published to RabbitMQ!")

    connection.close()
```

```python
# COMMAND: Create RabbitMQ consumer

def process_message(ch, method, properties, body):

    message = json.loads(body.decode())

    print("\n📩 Message received from RabbitMQ!")

    print("Event:", message["event"])
    print("Data:", message["data"])

    print("✅ Message processed!")

    ch.basic_ack(
        delivery_tag=method.delivery_tag
    )
```

```python
# COMMAND: Start RabbitMQ consumer in background

def start_consumer():

    connection = create_rabbitmq_connection()

    channel = connection.channel()

    channel.queue_declare(
        queue=QUEUE_NAME,
        durable=True
    )

    channel.basic_consume(
        queue=QUEUE_NAME,
        on_message_callback=process_message
    )

    print("👂 RabbitMQ consumer started!")
    print("Waiting for messages...")

    channel.start_consuming()


consumer_thread = threading.Thread(
    target=start_consumer,
    daemon=True
)

consumer_thread.start()

print("✅ Consumer running in background!")
```

```python
# COMMAND: Create feature engineering function

def create_features(data):

    df = pd.DataFrame([data])

    df["years_since_graduation"] = (
        2026 - df["graduation_year"]
    )

    df["email_open_rate"] = (
        df["emails_opened"] /
        df["emails_received"]
    ).fillna(0).clip(0, 1)

    df["average_donation_amount"] = (
        df["total_donation_amount"] /
        df["previous_donations"]
    ).fillna(0)

    df["donation_recency_score"] = (
        1 / (1 + df["days_since_last_donation"])
    )

    df["event_engagement"] = (
        df["events_attended"] /
        df["events_attended"].max()
    )

    df["newsletter_engagement"] = (
        df["newsletter_clicks"] /
        df["newsletter_clicks"].max()
    )

    df["volunteer_engagement"] = (
        df["volunteer_events"] /
        df["volunteer_events"].max()
    )

    df["interaction_recency"] = (
        1 -
        df["days_since_last_interaction"] /
        df["days_since_last_interaction"].max()
    ).clip(0, 1)

    df["engagement_score"] = (
        0.30 * df["event_engagement"]
        + 0.30 * df["email_open_rate"]
        + 0.15 * df["newsletter_engagement"]
        + 0.15 * df["volunteer_engagement"]
        + 0.10 * df["interaction_recency"]
    ) * 100

    return df

print("✅ Feature engineering ready!")
```

```python
# COMMAND: Create feature engineering function

def create_features(data):

    df = pd.DataFrame([data])

    df["years_since_graduation"] = (
        2026 - df["graduation_year"]
    )

    df["email_open_rate"] = (
        df["emails_opened"] /
        df["emails_received"]
    ).fillna(0).clip(0, 1)

    df["average_donation_amount"] = (
        df["total_donation_amount"] /
        df["previous_donations"]
    ).fillna(0)

    df["donation_recency_score"] = (
        1 / (1 + df["days_since_last_donation"])
    )

    df["event_engagement"] = (
        df["events_attended"] /
        df["events_attended"].max()
    )

    df["newsletter_engagement"] = (
        df["newsletter_clicks"] /
        df["newsletter_clicks"].max()
    )

    df["volunteer_engagement"] = (
        df["volunteer_events"] /
        df["volunteer_events"].max()
    )

    df["interaction_recency"] = (
        1 -
        df["days_since_last_interaction"] /
        df["days_since_last_interaction"].max()
    ).clip(0, 1)

    df["engagement_score"] = (
        0.30 * df["event_engagement"]
        + 0.30 * df["email_open_rate"]
        + 0.15 * df["newsletter_engagement"]
        + 0.15 * df["volunteer_engagement"]
        + 0.10 * df["interaction_recency"]
    ) * 100

    return df

print("✅ Feature engineering ready!")
```
